```python
import os, json
```

```python
# 원본 JSON 폴더와 새로 저장할 폴더 경로
input_folder = r"C:/Users/User/Desktop/fire_hydrant/labels"
output_folder = r"C:/Users/User/Desktop/fire_hydrant/clean_labels"
```

```python
# 라벨링 데이터 정제
count = 0
for filename in os.listdir(input_folder):
    if filename.endswith(".json"):
        input_path = os.path.join(input_folder, filename)
        output_path = os.path.join(output_folder, filename)

        # JSON 파일 열기
        with open(input_path, 'r', encoding='utf-8') as infile:
            data = json.load(infile)

        # 정제될 구조 생성
        cleaned_data = {
            "info": {
                "image_id": data["info"].get("image_id"),
                "filename": data["info"].get("filename"),
                "width": data["info"].get("width"),
                "height": data["info"].get("height")
            },
            "annotations": []
        }

        for ann in data.get("annotations", []):
            cleaned_ann = {
                "annotation_id": ann.get("annotation_id"),
                "label_name": "fire_hydrant",   # 라벨링 이름
                "annotation_type": ann.get("annotation_type"),
                "annotation_info": ann.get("annotation_info")
            }
            cleaned_data["annotations"].append(cleaned_ann)

        # 정제된 JSON 저장
        with open(output_path, 'w', encoding='utf-8') as outfile:
            json.dump(cleaned_data, outfile, ensure_ascii=False, indent=4)
    count += 1
```

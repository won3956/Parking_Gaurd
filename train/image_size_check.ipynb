```python
from PIL import Image
import pillow_heif  # HEIC 지원
import os

# HEIC 등록 (한 번만 실행해도 됨)
pillow_heif.register_heif_opener()

def resize_all_images(input_dir, output_dir, target_size=(4032, 3024)):
    os.makedirs(output_dir, exist_ok=True)
    supported_ext = ('.jpg', '.jpeg', '.png', '.mpo', '.heic')

    for fname in sorted(os.listdir(input_dir)):
        if not fname.lower().endswith(supported_ext):
            continue

        input_path = os.path.join(input_dir, fname)
        output_path = os.path.join(output_dir, os.path.splitext(fname)[0] + '.jpg')  # 전부 JPEG 저장

        try:
            with Image.open(input_path) as img:
                resized = img.resize(target_size, Image.LANCZOS)
                resized = resized.convert("RGB")  # JPEG 저장을 위해 필요
                resized.save(output_path, format='JPEG', quality=95)
                print(f"✅ {fname} → 리사이즈 완료 (JPEG)")
        except Exception as e:
            print(f"❌ {fname} 처리 실패: {e}")
```

```python
resize_all_images('C:/Users/User/Downloads/dataset/night', 'C:/Users/User/Downloads/dataset/1')
```

```python
# 이미지 잘 변환되었는지 확인해보는 코드

from PIL import Image
import os

def print_image_properties(image_dir):
    print(f"{'파일명':<30} {'해상도':<15} {'모드':<10} {'포맷':<10} {'크기(Byte)':<12}")
    print("-" * 80)

    for fname in sorted(os.listdir(image_dir)):
        if not fname.lower().endswith(('.jpg', '.jpeg', '.png')):
            continue

        fpath = os.path.join(image_dir, fname)
        try:
            with Image.open(fpath) as img:
                resolution = img.size  # (width, height)
                mode = img.mode
                fmt = img.format
                fsize = os.path.getsize(fpath)

                print(f"{fname:<30} {str(resolution):<15} {mode:<10} {fmt:<10} {fsize:<12}")
        except Exception as e:
            print(f"{fname:<30} ❌ 읽기 실패: {e}")

print_image_properties('C:/Users/User/Downloads/dataset/images')
```

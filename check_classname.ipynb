```python
#!/usr/bin/env python
# -*- coding: utf-8 -*-
"""
폴더 내 *.json 라벨 파일에서
- 'crosswalkd'  →  'crosswalk' 로 교정
- 교정된 개수 출력
- 데이터셋 전체 클래스 목록 출력
"""

import json
from pathlib import Path

LABEL_DIR = Path("C:/Users/User/Downloads/dataset/labels")          # 라벨 json 폴더
TYPO      = "crosswalkd"            # 잘못된 클래스명
CORRECT   = "crosswalk"             # 올바른 클래스명

fixed_count  = 0                    # 교정된 shape 수
all_classes  = set()                # 전체 클래스 모음

for json_path in LABEL_DIR.glob("*.json"):
    with open(json_path, "r", encoding="utf-8") as f:
        data = json.load(f)

    modified = False
    for shape in data.get("shapes", []):
        label = shape.get("label")
        all_classes.add(label)

        if label == TYPO:
            shape["label"] = CORRECT
            modified = True
            fixed_count += 1
            # 집계용 set에도 교정된 이름 반영
            all_classes.discard(TYPO)
            all_classes.add(CORRECT)

    # 교정이 있었으면 파일 덮어쓰기
    if modified:
        with open(json_path, "w", encoding="utf-8") as f:
            json.dump(data, f, ensure_ascii=False, indent=2)

# ――― 결과 ―――
print(f"교정 완료: '{TYPO}' → '{CORRECT}'  →  {fixed_count} 개 수정")
print("현재 데이터셋 클래스 목록:")
for c in sorted(all_classes):
    print("  -", c)
```
